```python
import pandas as pd
import numpy as np
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.width", None)
```

```python
df = pd.read_parquet("../data/raw/electricity_raw.parquet", engine="fastparquet")

# Verify load
df.head()
```

```python
# --- Dataset Overview ---------------------------------------------
rows, cols = (df.shape)
print(f"Rows: {rows}")
print(f"Columns: {cols}")
```

```python
df.info()
```

```python
# --- Descriptive Stats ---------------------------------------------
df.describe()
```

```python
print("Missing values for each column")
print(df.isna().sum())
```

```python
# Duplicated rows / timestamps
print(f"Duplicated rows: {df.duplicated().sum()}")
print(f"Duplicated timestamp: {df["timestamp"].duplicated().sum()}")
```

```python
# --- Numerical Distributions ---------------------------------------------

numeric_columns = df.select_dtypes(
    include=np.number
).columns

for column in numeric_columns:
    fig = px.histogram(
        df,
        x=column,
        nbins=30,
        title=f"Distribution of {column}",
        labels={column: column},
        width=800,
        height=400
    )

    fig.update_layout(
        xaxis_title=column,
        yaxis_title="Frequency"
    )

    fig.write_html(f"../figures/histograms/{column}_distribution.html")
```

```python
# --- Outlier Detection ---------------------------------------------

for column in numeric_columns:
    fig = px.box(
        df,
        x=column,
        title=f"Boxplot of {column}"
    )

    fig.write_html(f"../figures/boxplots/{column}_boxplot.html")
```
