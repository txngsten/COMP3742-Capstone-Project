```python
import pandas as pd
import numpy as np
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
pd.set_option("display.width", None)
```

```python
df = pd.read_parquet("../data/raw/electricity_raw.parquet", engine="fastparquet")

# Verify load
df.head()
```

```python
# --- Dataset Overview ---------------------------------------------
rows, cols = (df.shape)
print(f"Rows: {rows}")
print(f"Columns: {cols}")
```

```python
# --- Descriptive Stats ---------------------------------------------
df.describe()
```

```python
# Duplicated rows / timestamps
print(f"Duplicated rows: {df.duplicated().sum()}")
print(f"Duplicated timestamp: {df["timestamp"].duplicated().sum()}")
```

```python
# --- Coverage Investigation  ---------------------------------------------
ts = pd.DatetimeIndex(df["timestamp"]).sort_values()

expected = pd.date_range(ts.min(), ts.max(), freq="h")
missing_hours = expected.difference(ts)

print("Expected hourly rows:", len(expected))
print("Missing hours:", len(missing_hours))
display(missing_hours[:48])
```

```python
# --- Numerical Distributions ---------------------------------------------

numeric_columns = df.select_dtypes(
    include=np.number
).columns

for column in numeric_columns:
    fig = px.histogram(
        df,
        x=column,
        nbins=30,
        title=f"Distribution of {column}",
        labels={column: column},
        width=800,
        height=400
    )

    fig.update_layout(
        xaxis_title=column,
        yaxis_title="Frequency"
    )

    fig.write_html(f"../figures/histograms/{column}_distribution.html")
```

```python
# --- Outlier Detection ---------------------------------------------

for column in numeric_columns:
    fig = px.box(
        df,
        x=column,
        title=f"Boxplot of {column}"
    )

    fig.write_html(f"../figures/boxplots/{column}_boxplot.html")
```

```python
negative_curtailment = df[df["curtailment"] < 0]

print("Negative values:", len(negative_curtailment))
print(negative_curtailment["curtailment"].describe())
print(negative_curtailment["curtailment"].sort_values().head(20))

fig = px.histogram(
    df[df["curtailment"] < 0],
    x="curtailment",
    nbins=50,
    title="Distribution of Negative Curtailment Values"
)

fig.show()
```

```python
df.loc[
    df["curtailment"] < -10,
    ["timestamp", "curtailment"]
].sort_values("curtailment")
```

```python
# Keep original value
df["curtailment_raw"] = df["curtailment"]

# Physical interpretation of actual curtailed generation
df["curtailment_positive"] = df["curtailment"].clip(lower=0)

# Record whether the original calculation was negative
df["negative_curtailment"] = (
    df["curtailment"] < 0
).astype(int)
```

The curtailment variable contained 1,128 negative observations. Although most negative values were relatively small (median = -0.852 MW), several substantially negative observations were present, with a minimum of -65.397 MW. As curtailment represents generation that could have occurred but was not dispatched, negative observations were treated as potentially invalid and investigated before cleaning. Decided not to remove negative values, but instead, split the column into 3 categories/interpretations which can be used later on.
